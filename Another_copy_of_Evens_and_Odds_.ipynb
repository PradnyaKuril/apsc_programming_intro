## Example 8 — Guess Number 1–100

| | Task |
|---|---|
| 🖥️ **BCA** | Code a random guess. Then code a halving strategy. Which is smarter? |
| 📊 **BBA** | Shortlist a vendor from 10 options. Do you check randomly or systematically? |

> 🔍 *Common thinking gap:* Random vs systematic binary search — reveals thinking style

```python
# Example 8 — Guess a Number
import random
secret  = random.randint(1, 1000)
guesses = 0

# Strategy A: pure random
guess = random.randint(1, 1000)
guesses += 1
print(f"Random guess: {guess} | Secret was: {secret} | Tries: {guesses}")

# Strategy B: Binary search (smarter!)
low, high = 1, 1000
tries = 0
while low <= high:
    mid = (low + high) // 2
    tries += 1
    if mid == secret:
        print(f"Binary search found {secret} in {tries} tries!")
        break
    elif mid < secret:
        low = mid + 1
    else:
        high = mid - 1

# ── YOUR TASK ──────────────────────────────────────────
# BCA: Change the range to 1–1000. How many tries does binary search need?
# BBA: What is the maximum number of tries Strategy B ever needs for 100 items?
```
